```python
import pandas as pd
import os
```

```python
input_file = "D:\\Magang\\Projek\\Review Prosedur\\dokumen\\data-review-prosedur.xlsx"
output_folder = "../dokumen"

os.makedirs(output_folder, exist_ok=True)
```

```python
df = pd.read_excel(input_file)

# Pastikan kolom PIC ada
if "PIC" not in df.columns:
    raise ValueError("Kolom 'PIC' tidak ditemukan di file Excel")

grouped = df.groupby("PIC")

for pic_name, data_pic in grouped:
    # Bersihkan nama PIC agar aman untuk nama file
    safe_pic_name = str(pic_name).replace("/", "_").replace("\\", "_")
    
    output_file = os.path.join(
        output_folder,
        f"dokumen_{safe_pic_name}.xlsx"
    )
    
    # Simpan ke Excel
    data_pic.to_excel(output_file, index=False)
    
    print(f"File dibuat: {output_file}")

print("Proses selesai 🎉")
```
